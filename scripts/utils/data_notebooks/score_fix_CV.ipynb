```python
import pandas as pd
```

```python
new_scores = pd.read_csv("/users/1/lundq163/projects/automated-qc/data/known_problems/re-eval_for_CV_data_model_02r7.csv")
new_scores
```

```python
# drop the following columns: 'QU_motion', 'Rater', 'Notes', 'accuracy'
#new_scores = new_scores.drop(columns=['QU_motion', 'Rater', 'Notes', 'accuracy'])
new_scores = new_scores[['scan', 'Updated QU motion score']]
new_scores
```

```python
# load in current data set
#current_scores = pd.read_csv("/users/1/lundq163/projects/automated-qc/data/raw/auto_qc_t1w_t2w_subset_1024r.csv")
current_scores = pd.read_csv("/users/1/lundq163/projects/automated-qc/data/raw/auto_qc_t1w_t2w_subset_1024r_curated.csv")
current_scores
```

```python
# merge the two dataframes on the scan column
merged_scores = current_scores.merge(new_scores, on='scan', how='left')
merged_scores
```

```python
# rename the 'Updated QU motion score' column to 'corrected_QU_motion'
merged_scores = merged_scores.rename(columns={'Updated QU motion score': 'corrected_QU_motion'})
merged_scores
```

```python
# if there are any missing values in the 'corrected_QU_motion' column, fill them with the original 'QU_motion' scores
merged_scores['corrected_QU_motion'] = merged_scores['corrected_QU_motion'].fillna(merged_scores['QU_motion'])
merged_scores
```

```python
# drop rows where corrected_QU_motion is NaN
merged_scores = merged_scores.dropna(subset=['corrected_QU_motion'])
merged_scores
```

```python
# save as new csv
merged_scores.to_csv("/users/1/lundq163/projects/automated-qc/data/raw/auto_qc_t1w_t2w_subset_1024r_fixed_from_model02r7_corrections.csv", index=False)
```
